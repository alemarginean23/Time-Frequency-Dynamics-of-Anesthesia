```python
import os
import shutil
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.signal import spectrogram
from sklearn.model_selection import StratifiedKFold, train_test_split
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import confusion_matrix, f1_score

import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (Input, Conv2D, MaxPooling2D, Dense, Dropout, 
                                     BatchNormalization, GlobalAveragePooling2D)
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from tensorflow.keras.regularizers import l2
from tensorflow.keras.optimizers import Adam

plt.rcParams.update({
    'font.size': 18,
    'axes.titlesize': 20,
    'axes.labelsize': 18,
    'xtick.labelsize': 16,
    'ytick.labelsize': 16,
    'legend.fontsize': 16,
    'figure.titlesize': 22
})

# configurari generale
CLASS_NAMES = ["Deep", "Intermediate", "Light"]

DATASET_PATHS = {
    "0002": "Anestezie/M014_S001_SRCS3L_25,50,100_0002",  # deep (label 0)
    "0003": "Anestezie/M014_S001_SRCS3L_25,50,100_0003",  # intermediate (label 1)
    "0004": "Anestezie/M014_S001_SRCS3L_25,50,100_0004"   # light (label 2)
}

DATASET_LABELS = {
    "0002": 0,
    "0003": 1,
    "0004": 2
}

SPECTROGRAMS_DIR = "Spectrograms"
EXCLUDED_CHANNELS = {1, 22, 33}

SAMPLE_RATE = 1000
DURATION_SEC = 120                          
TOTAL_SAMPLES = DURATION_SEC * SAMPLE_RATE  

WINDOW_SEC = 2.0                            
STEP_SEC = 0.4                              

WINDOW_SAMPLES = int(WINDOW_SEC * SAMPLE_RATE)  
STEP_SAMPLES = int(STEP_SEC * SAMPLE_RATE)      

def compute_spectrogram(trial_data, fs=SAMPLE_RATE, nperseg=256, noverlap=128, max_freq=300):
    spec_list = []
    for ch in range(trial_data.shape[0]):
        f, t, Sxx = spectrogram(trial_data[ch, :], fs=fs, nperseg=nperseg, noverlap=noverlap)
        freq_mask = f <= max_freq
        spec_list.append(Sxx[freq_mask, :])
    return np.stack(spec_list, axis=-1)

def generate_sliding_window_data():
    common_channels = None
    for path in DATASET_PATHS.values():
        files = [f for f in os.listdir(path) if f.endswith(".bin") and "-Ch" in f]
        ch_ids = set([int(f.split("-Ch")[1].split(".")[0]) for f in files])
        if common_channels is None:
            common_channels = ch_ids
        else:
            common_channels = common_channels.intersection(ch_ids)
            
    valid_channels = sorted([ch for ch in common_channels if ch not in EXCLUDED_CHANNELS])
    num_channels = len(valid_channels)
    print(f"canale eliminate: {sorted(list(EXCLUDED_CHANNELS))}")
    print(f"canale retinute ({num_channels} canale): {valid_channels}")

    for dataset_id, path in DATASET_PATHS.items():
        print(f"\nprocesare sliding window pentru dataset {dataset_id}...")
        save_dir = os.path.join(SPECTROGRAMS_DIR, dataset_id)
        os.makedirs(save_dir, exist_ok=True)

        bin_files = sorted(
            [f for f in os.listdir(path) if f.endswith(".bin") and "-Ch" in f and int(f.split("-Ch")[1].split(".")[0]) in valid_channels],
            key=lambda x: int(x.split("-Ch")[1].split(".")[0])
        )

        raw_channels = []
        for file_name in bin_files:
            ch_data = np.memmap(os.path.join(path, file_name), dtype=np.float32, mode='r')
            raw_channels.append(ch_data[:TOTAL_SAMPLES])

        raw_matrix = np.array(raw_channels)

        trial_idx = 0
        start = 0
        while start + WINDOW_SAMPLES <= TOTAL_SAMPLES:
            end = start + WINDOW_SAMPLES
            trial_data = raw_matrix[:, start:end]

            spec = compute_spectrogram(trial_data)
            spec_native = np.log1p(spec).astype(np.float32)

            for ch_idx, physical_ch in enumerate(valid_channels):
                np.save(os.path.join(save_dir, f"channel_{physical_ch}_trial_{trial_idx}.npy"), spec_native[:, :, ch_idx])

            start += STEP_SAMPLES
            trial_idx += 1

        print(f"dataset {dataset_id}: au fost generate {trial_idx} trial-uri.")

def spec_augment(spectrogram, time_mask_param=2, freq_mask_param=5):
    aug_spec = np.copy(spectrogram)
    freq_bins, time_steps, channels = aug_spec.shape

    max_t = min(time_mask_param, time_steps)
    if max_t > 0:
        t_mask = np.random.randint(1, max_t + 1)
        t_0 = np.random.randint(0, time_steps - t_mask + 1)
        aug_spec[:, t_0:t_0 + t_mask, :] = 0.0

    max_f = min(freq_mask_param, freq_bins)
    if max_f > 0:
        f_mask = np.random.randint(1, max_f + 1)
        f_0 = np.random.randint(0, freq_bins - f_mask + 1)
        aug_spec[f_0:f_0 + f_mask, :, :] = 0.0

    return aug_spec

def load_precomputed_data(base_dir=SPECTROGRAMS_DIR):
    X, y = [], []
    for dataset_id, label in DATASET_LABELS.items():
        ds_path = os.path.join(base_dir, dataset_id)
        if not os.path.exists(ds_path):
            continue
            
        files = os.listdir(ds_path)
        channels, trials = set(), set()
        for f in files:
            if f.startswith("channel_") and f.endswith(".npy"):
                parts = f.replace(".npy", "").split("_")
                channels.add(int(parts[1]))
                trials.add(int(parts[3]))
                
        sorted_channels = sorted(list(channels))
        sorted_trials = sorted(list(trials))
        
        if not sorted_trials:
            continue
            
        for trial in sorted_trials:
            trial_channels = []
            expected_shape = None
            for ch in sorted_channels:
                file_path = os.path.join(ds_path, f"channel_{ch}_trial_{trial}.npy")
                spec = np.load(file_path)
                if expected_shape is None:
                    expected_shape = spec.shape
                trial_channels.append(spec)
            
            trial_tensor = np.stack(trial_channels, axis=-1)
            X.append(trial_tensor)
            y.append(label)

    X = np.array(X, dtype=np.float32)
    y = np.array(y)
    return X, y

def build_cnn_model(input_shape):
    inputs = Input(shape=input_shape)
    reg = l2(0.005)

    x = Conv2D(8, (3, 3), activation='relu', padding='same', kernel_regularizer=reg)(inputs)
    x = BatchNormalization()(x)
    x = MaxPooling2D((2, 2))(x) 
    x = Dropout(0.3)(x)

    x = Conv2D(16, (3, 3), activation='relu', padding='same', kernel_regularizer=reg, name='xai_target_conv')(x)
    x = BatchNormalization()(x)
    x = MaxPooling2D((2, 2))(x) 
    x = Dropout(0.4)(x)

    x = GlobalAveragePooling2D()(x) 
    x = Dense(16, activation='relu', kernel_regularizer=reg)(x)
    x = Dropout(0.4)(x)
    
    outputs = Dense(len(CLASS_NAMES), activation='softmax')(x)

    model = Model(inputs, outputs)
    model.compile(optimizer=Adam(learning_rate=0.0005), loss='sparse_categorical_crossentropy', metrics=['accuracy'])
    return model

def train_model(X, y):
    # hold-out 20% inainte de cross-validation
    X_train_val, X_test_extern, y_train_val, y_test_extern = train_test_split(
        X, y, test_size=0.20, stratify=y, random_state=42
    )
    
    print(f"Total date in dataset: {len(X)}")
    print(f"Date rezervate pentru Setul de Test Extern (hold-out, neatinse in CV): {len(X_test_extern)}")
    print(f"Date ramase pentru 3-Fold Cross-Validation: {len(X_train_val)}")

    skf = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
    fold_accuracies = []
    fold_f1_scores = []

    best_model = None
    best_f1 = -1
    N_AUG = 2  

    for fold, (train_idx, val_idx) in enumerate(skf.split(X_train_val, y_train_val)):
        print(f"\npornire Fold {fold + 1} ------------------------------")

        X_train, X_val = X_train_val[train_idx], X_train_val[val_idx]
        y_train, y_val = y_train_val[train_idx], y_train_val[val_idx]

        train_mean = np.mean(X_train)
        train_std = np.std(X_train)
        
        X_train = (X_train - train_mean) / (train_std + 1e-8)
        X_val = (X_val - train_mean) / (train_std + 1e-8)

        X_train_aug_list = []
        y_train_aug_list = []

        for _ in range(N_AUG):
            X_batch = np.array([spec_augment(spec, time_mask_param=2, freq_mask_param=5) for spec in X_train])
            X_train_aug_list.append(X_batch)
            y_train_aug_list.append(y_train)

        if N_AUG > 0:
            X_train = np.concatenate([X_train] + X_train_aug_list, axis=0)
            y_train = np.concatenate([y_train] + y_train_aug_list, axis=0)

        shuffle_idx = np.random.permutation(len(X_train))
        X_train = X_train[shuffle_idx]
        y_train = y_train[shuffle_idx]

        model = build_cnn_model(X_train.shape[1:])
        
        callbacks = [
            EarlyStopping(monitor='val_loss', patience=15, restore_best_weights=True),
            ReduceLROnPlateau(monitor='val_loss', factor=0.3, patience=5, min_lr=1e-7)
        ]

        class_weights = compute_class_weight('balanced', classes=np.unique(y_train), y=y_train)
        class_weight_dict = dict(zip(np.unique(y_train), class_weights))

        history = model.fit(
            X_train, y_train,
            epochs=80,
            batch_size=32, 
            validation_data=(X_val, y_val),
            class_weight=class_weight_dict,
            callbacks=callbacks,
            verbose=1
        )

        y_pred = np.argmax(model.predict(X_val), axis=1)
        acc = np.mean(y_pred == y_val)
        f1 = f1_score(y_val, y_pred, average='weighted')

        fold_accuracies.append(acc)
        fold_f1_scores.append(f1)

        if f1 > best_f1:
            best_f1 = f1
            best_model = model

        print(f"\nFold {fold + 1} Acuratete (Val): {acc:.4f} | F1: {f1:.4f}")

        # --- AFISARE MATRICE DE CONFUZIE PENTRU FIECARE FOLD ---
        cm = confusion_matrix(y_val, y_pred)
        plt.figure(figsize=(8, 6))
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                    xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES, annot_kws={"size": 16})
        plt.xlabel("Predicted")
        plt.ylabel("True")
        plt.title(f"Confusion Matrix - Fold {fold + 1} (Validation)")
        plt.tight_layout()
        plt.show()

        # --- AFISARE GRAFICE ACCURACY & LOSS PENTRU FIECARE FOLD ---
        plt.figure(figsize=(14, 5))
        plt.subplot(1, 2, 1)
        plt.plot(history.history['accuracy'], label='Train')
        plt.plot(history.history['val_accuracy'], label='Validation')
        plt.title(f'Accuracy - Fold {fold + 1}')
        plt.xlabel('Epoch')
        plt.ylabel('Accuracy')
        plt.legend()

        plt.subplot(1, 2, 2)
        plt.plot(history.history['loss'], label='Train')
        plt.plot(history.history['val_loss'], label='Validation')
        plt.title(f'Loss - Fold {fold + 1}')
        plt.xlabel('Epoch')
        plt.ylabel('Loss')
        plt.legend()
        plt.tight_layout()
        plt.show()

    # --- TABEL CU REZULTATE PE FOLDURI ---
    print("\n" + "="*50)
    print("REZULTATE 3-FOLD CROSS-VALIDATION:")
    print("="*50)
    print(f"{'Fold':<10} | {'Acuratete (Val)':<18} | {'F1-Score (Val)':<15}")
    print("-" * 50)
    for i, (acc, f1) in enumerate(zip(fold_accuracies, fold_f1_scores)):
        print(f"Fold {i + 1:<5} | {acc:.4f}           | {f1:.4f}")
    print("-" * 50)
    print(f"Medie CV   | {np.mean(fold_accuracies):.4f} ± {np.std(fold_accuracies):.4f} | {np.mean(fold_f1_scores):.4f} ± {np.std(fold_f1_scores):.4f}")
    print("="*50)

    # EVALUAREA FINALA PE SETUL DE TEST EXTERN (HOLD-OUT)
    print("\nEvaluare finala pe Setul de Test Extern (Hold-out complet nevazut)...")
    mean_ext = np.mean(X_train_val)
    std_ext = np.std(X_train_val)
    X_test_extern_norm = (X_test_extern - mean_ext) / (std_ext + 1e-8)

    y_pred_extern = np.argmax(best_model.predict(X_test_extern_norm), axis=1)
    test_acc = np.mean(y_pred_extern == y_test_extern)
    test_f1 = f1_score(y_test_extern, y_pred_extern, average='weighted')

    print(f"\n✨ REZULTAT FINAL PE SETUL DE TEST EXTERN:")
    print(f"   Acuratete Test Extern = {test_acc:.4f}")
    print(f"   Scor F1 Test Extern     = {test_f1:.4f}")

    # Matrice de confuzie pe testul extern
    cm_extern = confusion_matrix(y_test_extern, y_pred_extern)
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm_extern, annot=True, fmt='d', cmap='Greens',
                xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES, annot_kws={"size": 16})
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.title("Confusion Matrix - External Test Set (Hold-out Final)")
    plt.tight_layout()
    plt.show()

    return best_model

if __name__ == "__main__":
    if os.path.exists(SPECTROGRAMS_DIR):
        print("stergem folderul vechi 'Spectrograms'...")
        shutil.rmtree(SPECTROGRAMS_DIR)

    print("generam spectrogramele native <= 300 Hz cu sliding window...")
    generate_sliding_window_data()

    X, y = load_precomputed_data()
    
    if len(X) > 0:
        best_model = train_model(X, y)
        best_model.save("cnn_augmentation_step0.4.h5")
        print("\nmodelul optim a fost salvat in cnn_augmentation_step0.4.h5")
    else:
        print("nu s-au putut incarca datele.")
```

```python
import os
import shutil
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.signal import spectrogram
from sklearn.model_selection import StratifiedKFold
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import confusion_matrix, f1_score

import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (Input, Conv2D, MaxPooling2D, Dense, Dropout, 
                                     BatchNormalization, GlobalAveragePooling2D)
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from tensorflow.keras.regularizers import l2
from tensorflow.keras.optimizers import Adam

plt.rcParams.update({
    'font.size': 18,
    'axes.titlesize': 20,
    'axes.labelsize': 18,
    'xtick.labelsize': 16,
    'ytick.labelsize': 16,
    'legend.fontsize': 16,
    'figure.titlesize': 22
})

# configurari generale
CLASS_NAMES = ["Deep", "Intermediate", "Light"]

DATASET_PATHS = {
    "0002": "Anestezie/M014_S001_SRCS3L_25,50,100_0002",  # deep (label 0)
    "0003": "Anestezie/M014_S001_SRCS3L_25,50,100_0003",  # intermediate (label 1)
    "0004": "Anestezie/M014_S001_SRCS3L_25,50,100_0004"   # light (label 2)
}

DATASET_LABELS = {
    "0002": 0,
    "0003": 1,
    "0004": 2
}

SPECTROGRAMS_DIR = "Spectrograms"

EXCLUDED_CHANNELS = {1, 22, 33}

#parametri semnal si sliding window
SAMPLE_RATE = 1000
DURATION_SEC = 120                          #primele 2 minute de activitate spontana
TOTAL_SAMPLES = DURATION_SEC * SAMPLE_RATE  #120.000 esantioane

WINDOW_SEC = 2.0                            #2 secunde per trial
#STEP_SEC = 0.4                              #pas de avansare de 0.4 secunde si asa avem overlap de 80%
#STEP_SEC = 0.2
STEP_SEC = 0.9

WINDOW_SAMPLES = int(WINDOW_SEC * SAMPLE_RATE)  #2000 esantioane (2 secunde x 1000hz sample freq)
STEP_SAMPLES = int(STEP_SEC * SAMPLE_RATE)      #400 esantioane

#spectrograma stft limitata la 0 - 300 Hz
def compute_spectrogram(trial_data, fs=SAMPLE_RATE, nperseg=256, noverlap=128, max_freq=300):
    spec_list = []
    for ch in range(trial_data.shape[0]):
        f, t, Sxx = spectrogram(trial_data[ch, :], fs=fs, nperseg=nperseg, noverlap=noverlap)
        
        #boolean mask ca sa am doar frecventele <= 300 Hz (77 benzi) 
        #o banda are 1000hz/256=3.906... si daca eu vreau sa limitez la 300hz atunci 300hz/3.990625hz/banda o sa am 77 benzi de freq
        freq_mask = f <= max_freq
        spec_list.append(Sxx[freq_mask, :])
        
    return np.stack(spec_list, axis=-1)


def generate_sliding_window_data():
    common_channels = None
    for path in DATASET_PATHS.values():
        files = [f for f in os.listdir(path) if f.endswith(".bin") and "-Ch" in f]
        ch_ids = set([int(f.split("-Ch")[1].split(".")[0]) for f in files])
        if common_channels is None:
            common_channels = ch_ids
        else:
            common_channels = common_channels.intersection(ch_ids)
            
    # eliminam canalele 1, 22 si 33
    valid_channels = sorted([ch for ch in common_channels if ch not in EXCLUDED_CHANNELS])
    num_channels = len(valid_channels)
    print(f"canale eliminate: {sorted(list(EXCLUDED_CHANNELS))}")
    print(f"canale retinute ({num_channels} canale): {valid_channels}")

    for dataset_id, path in DATASET_PATHS.items():
        print(f"\nprocesare sliding window pentru dataset {dataset_id}...")
        save_dir = os.path.join(SPECTROGRAMS_DIR, dataset_id)
        os.makedirs(save_dir, exist_ok=True)

        bin_files = sorted(
            [f for f in os.listdir(path) if f.endswith(".bin") and "-Ch" in f and int(f.split("-Ch")[1].split(".")[0]) in valid_channels],
            key=lambda x: int(x.split("-Ch")[1].split(".")[0])
        )

        raw_channels = []
        for file_name in bin_files:
            ch_data = np.memmap(os.path.join(path, file_name), dtype=np.float32, mode='r')
            raw_channels.append(ch_data[:TOTAL_SAMPLES])

        raw_matrix = np.array(raw_channels)

        trial_idx = 0
        start = 0
        while start + WINDOW_SAMPLES <= TOTAL_SAMPLES:
            end = start + WINDOW_SAMPLES
            trial_data = raw_matrix[:, start:end]

            spec = compute_spectrogram(trial_data)
            spec_native = np.log1p(spec).astype(np.float32)

            for ch_idx, physical_ch in enumerate(valid_channels):
                np.save(os.path.join(save_dir, f"channel_{physical_ch}_trial_{trial_idx}.npy"), spec_native[:, :, ch_idx])

            start += STEP_SAMPLES
            trial_idx += 1

        print(f"dataset {dataset_id}: au fost generate {trial_idx} trial-uri.")

# augmentare SpecAugment adaptata la dimensiunea (77, 14, 30)
def spec_augment(spectrogram, time_mask_param=2, freq_mask_param=5):
    aug_spec = np.copy(spectrogram)
    freq_bins, time_steps, channels = aug_spec.shape

    # 1. masca temporala: 1-2 pasi mascati din cei 14
    max_t = min(time_mask_param, time_steps)
    if max_t > 0:
        t_mask = np.random.randint(1, max_t + 1)
        t_0 = np.random.randint(0, time_steps - t_mask + 1)
        aug_spec[:, t_0:t_0 + t_mask, :] = 0.0

    # 2. masca de frecventa: 1-5 benzi mascate din cele 77
    max_f = min(freq_mask_param, freq_bins)
    if max_f > 0:
        f_mask = np.random.randint(1, max_f + 1)
        f_0 = np.random.randint(0, freq_bins - f_mask + 1)
        aug_spec[f_0:f_0 + f_mask, :, :] = 0.0

    return aug_spec

def normalize(X):
    print("normalizare semnal...")
    mean = np.mean(X)
    std = np.std(X)
    return (X - mean) / (std + 1e-8)

def load_precomputed_data(base_dir=SPECTROGRAMS_DIR):
    X, y = [], []
    
    for dataset_id, label in DATASET_LABELS.items():
        ds_path = os.path.join(base_dir, dataset_id)
        if not os.path.exists(ds_path):
            print(f"folder lipsa: {ds_path}")
            continue
            
        print(f"incarcare dataset {dataset_id}...")
        files = os.listdir(ds_path)
        
        channels = set()
        trials = set()
        for f in files:
            if f.startswith("channel_") and f.endswith(".npy"):
                parts = f.replace(".npy", "").split("_")
                channels.add(int(parts[1]))
                trials.add(int(parts[3]))
                
        sorted_channels = sorted(list(channels))
        sorted_trials = sorted(list(trials))
        
        if not sorted_trials:
            continue
            
        print(f"   gasit {len(sorted_trials)} trial-uri pe {len(sorted_channels)} canale.")
        
        for trial in sorted_trials:
            trial_channels = []
            expected_shape = None
            
            for ch in sorted_channels:
                file_path = os.path.join(ds_path, f"channel_{ch}_trial_{trial}.npy")
                spec = np.load(file_path)
                
                if expected_shape is None:
                    expected_shape = spec.shape
                elif spec.shape != expected_shape:
                    raise ValueError(f"shape mismatch in {dataset_id} la canal {ch}, trial {trial}")
                    
                trial_channels.append(spec)
            
            trial_tensor = np.stack(trial_channels, axis=-1)
            X.append(trial_tensor)
            y.append(label)

    X = np.array(X, dtype=np.float32)
    y = np.array(y)
    print(f"\ndimensiune finala tensor X (nativ <= 300 Hz): {X.shape}") 
    print(f"dimensiune etichete y: {y.shape}")
    return X, y

def build_cnn_model(input_shape):
    inputs = Input(shape=input_shape)
    reg = l2(0.005)

    x = Conv2D(8, (3, 3), activation='relu', padding='same', kernel_regularizer=reg)(inputs)
    x = BatchNormalization()(x)
    x = MaxPooling2D((2, 2))(x) 
    x = Dropout(0.3)(x)

    # target layer pentru xai (grad-cam)
    x = Conv2D(16, (3, 3), activation='relu', padding='same', kernel_regularizer=reg, name='xai_target_conv')(x)
    x = BatchNormalization()(x)
    x = MaxPooling2D((2, 2))(x) 
    x = Dropout(0.4)(x)

    x = GlobalAveragePooling2D()(x) 
    x = Dense(16, activation='relu', kernel_regularizer=reg)(x)
    x = Dropout(0.4)(x)
    
    outputs = Dense(len(CLASS_NAMES), activation='softmax')(x)

    model = Model(inputs, outputs)
    model.compile(optimizer=Adam(learning_rate=0.0005), loss='sparse_categorical_crossentropy', metrics=['accuracy'])
    return model

def train_model(X, y):
    # 1. Separare set de test extern (hold-out 20%) inainte de cross-validation
    X_train_val, X_test_extern, y_train_val, y_test_extern = train_test_split(
        X, y, test_size=0.20, stratify=y, random_state=42
    )
    
    print(f"Total date in dataset: {len(X)}")
    print(f"Date rezervate pentru Setul de Test Extern (hold-out, neatinse in CV): {len(X_test_extern)}")
    print(f"Date ramase pentru 3-Fold Cross-Validation: {len(X_train_val)}")

    skf = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
    fold_accuracies = []
    fold_f1_scores = []

    best_model = None
    best_f1 = -1
    N_AUG = 2  

    for fold, (train_idx, val_idx) in enumerate(skf.split(X_train_val, y_train_val)):
        print(f"\npornire Fold {fold + 1} ------------------------------")

        X_train, X_val = X_train_val[train_idx], X_train_val[val_idx]
        y_train, y_val = y_train_val[train_idx], y_train_val[val_idx]

        train_mean = np.mean(X_train)
        train_std = np.std(X_train)
        
        X_train = (X_train - train_mean) / (train_std + 1e-8)
        X_val = (X_val - train_mean) / (train_std + 1e-8)

        X_train_aug_list = []
        y_train_aug_list = []

        for _ in range(N_AUG):
            X_batch = np.array([spec_augment(spec, time_mask_param=2, freq_mask_param=5) for spec in X_train])
            X_train_aug_list.append(X_batch)
            y_train_aug_list.append(y_train)

        if N_AUG > 0:
            X_train = np.concatenate([X_train] + X_train_aug_list, axis=0)
            y_train = np.concatenate([y_train] + y_train_aug_list, axis=0)

        shuffle_idx = np.random.permutation(len(X_train))
        X_train = X_train[shuffle_idx]
        y_train = y_train[shuffle_idx]

        model = build_cnn_model(X_train.shape[1:])
        
        callbacks = [
            EarlyStopping(monitor='val_loss', patience=15, restore_best_weights=True),
            ReduceLROnPlateau(monitor='val_loss', factor=0.3, patience=5, min_lr=1e-7)
        ]

        class_weights = compute_class_weight('balanced', classes=np.unique(y_train), y=y_train)
        class_weight_dict = dict(zip(np.unique(y_train), class_weights))

        history = model.fit(
            X_train, y_train,
            epochs=80,
            batch_size=32, 
            validation_data=(X_val, y_val),
            class_weight=class_weight_dict,
            callbacks=callbacks,
            verbose=1
        )

        y_pred = np.argmax(model.predict(X_val), axis=1)
        acc = np.mean(y_pred == y_val)
        f1 = f1_score(y_val, y_pred, average='weighted')

        fold_accuracies.append(acc)
        fold_f1_scores.append(f1)

        if f1 > best_f1:
            best_f1 = f1
            best_model = model

        print(f"\nFold {fold + 1} Acuratete (Val): {acc:.4f} | F1: {f1:.4f}")

        # --- AFISARE MATRICE DE CONFUZIE PENTRU FIECARE FOLD ---
        cm = confusion_matrix(y_val, y_pred)
        plt.figure(figsize=(8, 6))
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                    xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES, annot_kws={"size": 16})
        plt.xlabel("Predicted")
        plt.ylabel("True")
        plt.title(f"Confusion Matrix - Fold {fold + 1} (Validation)")
        plt.tight_layout()
        plt.show()

        # --- AFISARE GRAFICE ACCURACY & LOSS PENTRU FIECARE FOLD ---
        plt.figure(figsize=(14, 5))
        plt.subplot(1, 2, 1)
        plt.plot(history.history['accuracy'], label='Train')
        plt.plot(history.history['val_accuracy'], label='Validation')
        plt.title(f'Accuracy - Fold {fold + 1}')
        plt.xlabel('Epoch')
        plt.ylabel('Accuracy')
        plt.legend()

        plt.subplot(1, 2, 2)
        plt.plot(history.history['loss'], label='Train')
        plt.plot(history.history['val_loss'], label='Validation')
        plt.title(f'Loss - Fold {fold + 1}')
        plt.xlabel('Epoch')
        plt.ylabel('Loss')
        plt.legend()
        plt.tight_layout()
        plt.show()

    # --- TABEL CU REZULTATE PE FOLDURI ---
    print("\n" + "="*50)
    print("REZULTATE 3-FOLD CROSS-VALIDATION:")
    print("="*50)
    print(f"{'Fold':<10} | {'Acuratete (Val)':<18} | {'F1-Score (Val)':<15}")
    print("-" * 50)
    for i, (acc, f1) in enumerate(zip(fold_accuracies, fold_f1_scores)):
        print(f"Fold {i + 1:<5} | {acc:.4f}           | {f1:.4f}")
    print("-" * 50)
    print(f"Medie CV   | {np.mean(fold_accuracies):.4f} ± {np.std(fold_accuracies):.4f} | {np.mean(fold_f1_scores):.4f} ± {np.std(fold_f1_scores):.4f}")
    print("="*50)

    # 3. EVALUAREA FINALA PE SETUL DE TEST EXTERN (HOLD-OUT)
    print("\nEvaluare finala pe Setul de Test Extern (Hold-out complet nevazut)...")
    mean_ext = np.mean(X_train_val)
    std_ext = np.std(X_train_val)
    X_test_extern_norm = (X_test_extern - mean_ext) / (std_ext + 1e-8)

    y_pred_extern = np.argmax(best_model.predict(X_test_extern_norm), axis=1)
    test_acc = np.mean(y_pred_extern == y_test_extern)
    test_f1 = f1_score(y_test_extern, y_pred_extern, average='weighted')

    print(f"\n✨ REZULTAT FINAL PE SETUL DE TEST EXTERN:")
    print(f"   Acuratete Test Extern = {test_acc:.4f}")
    print(f"   Scor F1 Test Extern     = {test_f1:.4f}")

    # Matrice de confuzie pe testul extern
    cm_extern = confusion_matrix(y_test_extern, y_pred_extern)
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm_extern, annot=True, fmt='d', cmap='Greens',
                xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES, annot_kws={"size": 16})
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.title("Confusion Matrix - External Test Set (Hold-out Final)")
    plt.tight_layout()
    plt.show()

    return best_model

# executie principala
if __name__ == "__main__":
    if os.path.exists(SPECTROGRAMS_DIR):
        print("stergem folderul vechi 'Spectrograms'...")
        shutil.rmtree(SPECTROGRAMS_DIR)

    print("generam spectrogramele native <= 300 Hz cu sliding window (2s fereastra, 0.4s/0.2s.2.0 pas)...")
    generate_sliding_window_data()

    X, y = load_precomputed_data()
    
    if len(X) > 0:
        best_model = train_model(X, y)
        best_model.save("cnn_augmentation_step0.9.h5")
        print("\nmodelul optim a fost salvat in cnn_augmentation_step0.9.h5")
    else:
        print("nu s-au putut incarca datele.")
```

```python
import os
import shutil
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.signal import spectrogram
from sklearn.model_selection import StratifiedKFold
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import confusion_matrix, f1_score

import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (Input, Conv2D, MaxPooling2D, Dense, Dropout, 
                                     BatchNormalization, GlobalAveragePooling2D)
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from tensorflow.keras.regularizers import l2
from tensorflow.keras.optimizers import Adam

plt.rcParams.update({
    'font.size': 18,
    'axes.titlesize': 20,
    'axes.labelsize': 18,
    'xtick.labelsize': 16,
    'ytick.labelsize': 16,
    'legend.fontsize': 16,
    'figure.titlesize': 22
})

# configurari generale
CLASS_NAMES = ["Deep", "Intermediate", "Light"]

DATASET_PATHS = {
    "0002": "Anestezie/M014_S001_SRCS3L_25,50,100_0002",  # deep (label 0)
    "0003": "Anestezie/M014_S001_SRCS3L_25,50,100_0003",  # intermediate (label 1)
    "0004": "Anestezie/M014_S001_SRCS3L_25,50,100_0004"   # light (label 2)
}

DATASET_LABELS = {
    "0002": 0,
    "0003": 1,
    "0004": 2
}

SPECTROGRAMS_DIR = "Spectrograms"

EXCLUDED_CHANNELS = {1, 22, 33}

#parametri semnal si sliding window
SAMPLE_RATE = 1000
DURATION_SEC = 120                          #primele 2 minute de activitate spontana
TOTAL_SAMPLES = DURATION_SEC * SAMPLE_RATE  #120.000 esantioane

WINDOW_SEC = 2.0                            #2 secunde per trial
#STEP_SEC = 0.4                              #pas de avansare de 0.4 secunde si asa avem overlap de 80%
#STEP_SEC = 0.2
STEP_SEC = 0.8

WINDOW_SAMPLES = int(WINDOW_SEC * SAMPLE_RATE)  #2000 esantioane (2 secunde x 1000hz sample freq)
STEP_SAMPLES = int(STEP_SEC * SAMPLE_RATE)      #400 esantioane

#spectrograma stft limitata la 0 - 300 Hz
def compute_spectrogram(trial_data, fs=SAMPLE_RATE, nperseg=256, noverlap=128, max_freq=300):
    spec_list = []
    for ch in range(trial_data.shape[0]):
        f, t, Sxx = spectrogram(trial_data[ch, :], fs=fs, nperseg=nperseg, noverlap=noverlap)
        
        #boolean mask ca sa am doar frecventele <= 300 Hz (77 benzi) 
        #o banda are 1000hz/256=3.906... si daca eu vreau sa limitez la 300hz atunci 300hz/3.990625hz/banda o sa am 77 benzi de freq
        freq_mask = f <= max_freq
        spec_list.append(Sxx[freq_mask, :])
        
    return np.stack(spec_list, axis=-1)


def generate_sliding_window_data():
    common_channels = None
    for path in DATASET_PATHS.values():
        files = [f for f in os.listdir(path) if f.endswith(".bin") and "-Ch" in f]
        ch_ids = set([int(f.split("-Ch")[1].split(".")[0]) for f in files])
        if common_channels is None:
            common_channels = ch_ids
        else:
            common_channels = common_channels.intersection(ch_ids)
            
    # eliminam canalele 1, 22 si 33
    valid_channels = sorted([ch for ch in common_channels if ch not in EXCLUDED_CHANNELS])
    num_channels = len(valid_channels)
    print(f"canale eliminate: {sorted(list(EXCLUDED_CHANNELS))}")
    print(f"canale retinute ({num_channels} canale): {valid_channels}")

    for dataset_id, path in DATASET_PATHS.items():
        print(f"\nprocesare sliding window pentru dataset {dataset_id}...")
        save_dir = os.path.join(SPECTROGRAMS_DIR, dataset_id)
        os.makedirs(save_dir, exist_ok=True)

        bin_files = sorted(
            [f for f in os.listdir(path) if f.endswith(".bin") and "-Ch" in f and int(f.split("-Ch")[1].split(".")[0]) in valid_channels],
            key=lambda x: int(x.split("-Ch")[1].split(".")[0])
        )

        raw_channels = []
        for file_name in bin_files:
            ch_data = np.memmap(os.path.join(path, file_name), dtype=np.float32, mode='r')
            raw_channels.append(ch_data[:TOTAL_SAMPLES])

        raw_matrix = np.array(raw_channels)

        trial_idx = 0
        start = 0
        while start + WINDOW_SAMPLES <= TOTAL_SAMPLES:
            end = start + WINDOW_SAMPLES
            trial_data = raw_matrix[:, start:end]

            spec = compute_spectrogram(trial_data)
            spec_native = np.log1p(spec).astype(np.float32)

            for ch_idx, physical_ch in enumerate(valid_channels):
                np.save(os.path.join(save_dir, f"channel_{physical_ch}_trial_{trial_idx}.npy"), spec_native[:, :, ch_idx])

            start += STEP_SAMPLES
            trial_idx += 1

        print(f"dataset {dataset_id}: au fost generate {trial_idx} trial-uri.")

# augmentare SpecAugment adaptata la dimensiunea (77, 14, 30)
def spec_augment(spectrogram, time_mask_param=2, freq_mask_param=5):
    aug_spec = np.copy(spectrogram)
    freq_bins, time_steps, channels = aug_spec.shape

    # 1. masca temporala: 1-2 pasi mascati din cei 14
    max_t = min(time_mask_param, time_steps)
    if max_t > 0:
        t_mask = np.random.randint(1, max_t + 1)
        t_0 = np.random.randint(0, time_steps - t_mask + 1)
        aug_spec[:, t_0:t_0 + t_mask, :] = 0.0

    # 2. masca de frecventa: 1-5 benzi mascate din cele 77
    max_f = min(freq_mask_param, freq_bins)
    if max_f > 0:
        f_mask = np.random.randint(1, max_f + 1)
        f_0 = np.random.randint(0, freq_bins - f_mask + 1)
        aug_spec[f_0:f_0 + f_mask, :, :] = 0.0

    return aug_spec

def normalize(X):
    print("normalizare semnal...")
    mean = np.mean(X)
    std = np.std(X)
    return (X - mean) / (std + 1e-8)

def load_precomputed_data(base_dir=SPECTROGRAMS_DIR):
    X, y = [], []
    
    for dataset_id, label in DATASET_LABELS.items():
        ds_path = os.path.join(base_dir, dataset_id)
        if not os.path.exists(ds_path):
            print(f"folder lipsa: {ds_path}")
            continue
            
        print(f"incarcare dataset {dataset_id}...")
        files = os.listdir(ds_path)
        
        channels = set()
        trials = set()
        for f in files:
            if f.startswith("channel_") and f.endswith(".npy"):
                parts = f.replace(".npy", "").split("_")
                channels.add(int(parts[1]))
                trials.add(int(parts[3]))
                
        sorted_channels = sorted(list(channels))
        sorted_trials = sorted(list(trials))
        
        if not sorted_trials:
            continue
            
        print(f"   gasit {len(sorted_trials)} trial-uri pe {len(sorted_channels)} canale.")
        
        for trial in sorted_trials:
            trial_channels = []
            expected_shape = None
            
            for ch in sorted_channels:
                file_path = os.path.join(ds_path, f"channel_{ch}_trial_{trial}.npy")
                spec = np.load(file_path)
                
                if expected_shape is None:
                    expected_shape = spec.shape
                elif spec.shape != expected_shape:
                    raise ValueError(f"shape mismatch in {dataset_id} la canal {ch}, trial {trial}")
                    
                trial_channels.append(spec)
            
            trial_tensor = np.stack(trial_channels, axis=-1)
            X.append(trial_tensor)
            y.append(label)

    X = np.array(X, dtype=np.float32)
    y = np.array(y)
    print(f"\ndimensiune finala tensor X (nativ <= 300 Hz): {X.shape}") 
    print(f"dimensiune etichete y: {y.shape}")
    return X, y

def build_cnn_model(input_shape):
    inputs = Input(shape=input_shape)
    reg = l2(0.005)

    x = Conv2D(8, (3, 3), activation='relu', padding='same', kernel_regularizer=reg)(inputs)
    x = BatchNormalization()(x)
    x = MaxPooling2D((2, 2))(x) 
    x = Dropout(0.3)(x)

    # target layer pentru xai (grad-cam)
    x = Conv2D(16, (3, 3), activation='relu', padding='same', kernel_regularizer=reg, name='xai_target_conv')(x)
    x = BatchNormalization()(x)
    x = MaxPooling2D((2, 2))(x) 
    x = Dropout(0.4)(x)

    x = GlobalAveragePooling2D()(x) 
    x = Dense(16, activation='relu', kernel_regularizer=reg)(x)
    x = Dropout(0.4)(x)
    
    outputs = Dense(len(CLASS_NAMES), activation='softmax')(x)

    model = Model(inputs, outputs)
    model.compile(optimizer=Adam(learning_rate=0.0005), loss='sparse_categorical_crossentropy', metrics=['accuracy'])
    return model

def train_model(X, y):
    # hold-out 20% inainte de cross-validation
    X_train_val, X_test_extern, y_train_val, y_test_extern = train_test_split(
        X, y, test_size=0.20, stratify=y, random_state=42
    )
    
    print(f"Total date in dataset: {len(X)}")
    print(f"Date rezervate pentru Setul de Test Extern (hold-out, neatinse in CV): {len(X_test_extern)}")
    print(f"Date ramase pentru 3-Fold Cross-Validation: {len(X_train_val)}")

    skf = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
    fold_accuracies = []
    fold_f1_scores = []

    best_model = None
    best_f1 = -1
    N_AUG = 2  

    for fold, (train_idx, val_idx) in enumerate(skf.split(X_train_val, y_train_val)):
        print(f"\npornire Fold {fold + 1} ------------------------------")

        X_train, X_val = X_train_val[train_idx], X_train_val[val_idx]
        y_train, y_val = y_train_val[train_idx], y_train_val[val_idx]

        train_mean = np.mean(X_train)
        train_std = np.std(X_train)
        
        X_train = (X_train - train_mean) / (train_std + 1e-8)
        X_val = (X_val - train_mean) / (train_std + 1e-8)

        X_train_aug_list = []
        y_train_aug_list = []

        for _ in range(N_AUG):
            X_batch = np.array([spec_augment(spec, time_mask_param=2, freq_mask_param=5) for spec in X_train])
            X_train_aug_list.append(X_batch)
            y_train_aug_list.append(y_train)

        if N_AUG > 0:
            X_train = np.concatenate([X_train] + X_train_aug_list, axis=0)
            y_train = np.concatenate([y_train] + y_train_aug_list, axis=0)

        shuffle_idx = np.random.permutation(len(X_train))
        X_train = X_train[shuffle_idx]
        y_train = y_train[shuffle_idx]

        model = build_cnn_model(X_train.shape[1:])
        
        callbacks = [
            EarlyStopping(monitor='val_loss', patience=15, restore_best_weights=True),
            ReduceLROnPlateau(monitor='val_loss', factor=0.3, patience=5, min_lr=1e-7)
        ]

        class_weights = compute_class_weight('balanced', classes=np.unique(y_train), y=y_train)
        class_weight_dict = dict(zip(np.unique(y_train), class_weights))

        history = model.fit(
            X_train, y_train,
            epochs=80,
            batch_size=32, 
            validation_data=(X_val, y_val),
            class_weight=class_weight_dict,
            callbacks=callbacks,
            verbose=1
        )

        y_pred = np.argmax(model.predict(X_val), axis=1)
        acc = np.mean(y_pred == y_val)
        f1 = f1_score(y_val, y_pred, average='weighted')

        fold_accuracies.append(acc)
        fold_f1_scores.append(f1)

        if f1 > best_f1:
            best_f1 = f1
            best_model = model

        print(f"\nFold {fold + 1} Acuratete (Val): {acc:.4f} | F1: {f1:.4f}")

        # --- AFISARE MATRICE DE CONFUZIE PENTRU FIECARE FOLD ---
        cm = confusion_matrix(y_val, y_pred)
        plt.figure(figsize=(8, 6))
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                    xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES, annot_kws={"size": 16})
        plt.xlabel("Predicted")
        plt.ylabel("True")
        plt.title(f"Confusion Matrix - Fold {fold + 1} (Validation)")
        plt.tight_layout()
        plt.show()

        # --- AFISARE GRAFICE ACCURACY & LOSS PENTRU FIECARE FOLD ---
        plt.figure(figsize=(14, 5))
        plt.subplot(1, 2, 1)
        plt.plot(history.history['accuracy'], label='Train')
        plt.plot(history.history['val_accuracy'], label='Validation')
        plt.title(f'Accuracy - Fold {fold + 1}')
        plt.xlabel('Epoch')
        plt.ylabel('Accuracy')
        plt.legend()

        plt.subplot(1, 2, 2)
        plt.plot(history.history['loss'], label='Train')
        plt.plot(history.history['val_loss'], label='Validation')
        plt.title(f'Loss - Fold {fold + 1}')
        plt.xlabel('Epoch')
        plt.ylabel('Loss')
        plt.legend()
        plt.tight_layout()
        plt.show()

    # --- TABEL CU REZULTATE PE FOLDURI ---
    print("\n" + "="*50)
    print("REZULTATE 3-FOLD CROSS-VALIDATION:")
    print("="*50)
    print(f"{'Fold':<10} | {'Acuratete (Val)':<18} | {'F1-Score (Val)':<15}")
    print("-" * 50)
    for i, (acc, f1) in enumerate(zip(fold_accuracies, fold_f1_scores)):
        print(f"Fold {i + 1:<5} | {acc:.4f}           | {f1:.4f}")
    print("-" * 50)
    print(f"Medie CV   | {np.mean(fold_accuracies):.4f} ± {np.std(fold_accuracies):.4f} | {np.mean(fold_f1_scores):.4f} ± {np.std(fold_f1_scores):.4f}")
    print("="*50)

    # EVALUAREA FINALA PE SETUL DE TEST EXTERN (HOLD-OUT)
    print("\nEvaluare finala pe Setul de Test Extern (Hold-out complet nevazut)...")
    mean_ext = np.mean(X_train_val)
    std_ext = np.std(X_train_val)
    X_test_extern_norm = (X_test_extern - mean_ext) / (std_ext + 1e-8)

    y_pred_extern = np.argmax(best_model.predict(X_test_extern_norm), axis=1)
    test_acc = np.mean(y_pred_extern == y_test_extern)
    test_f1 = f1_score(y_test_extern, y_pred_extern, average='weighted')

    print(f"\n✨ REZULTAT FINAL PE SETUL DE TEST EXTERN:")
    print(f"   Acuratete Test Extern = {test_acc:.4f}")
    print(f"   Scor F1 Test Extern     = {test_f1:.4f}")

    # Matrice de confuzie pe testul extern
    cm_extern = confusion_matrix(y_test_extern, y_pred_extern)
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm_extern, annot=True, fmt='d', cmap='Greens',
                xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES, annot_kws={"size": 16})
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.title("Confusion Matrix - External Test Set (Hold-out Final)")
    plt.tight_layout()
    plt.show()

    return best_model

# executie principala
if __name__ == "__main__":
    if os.path.exists(SPECTROGRAMS_DIR):
        print("stergem folderul vechi 'Spectrograms'...")
        shutil.rmtree(SPECTROGRAMS_DIR)

    print("generam spectrogramele native <= 300 Hz cu sliding window (2s fereastra, 0.4s/0.2s.2.0 pas)...")
    generate_sliding_window_data()

    X, y = load_precomputed_data()
    
    if len(X) > 0:
        best_model = train_model(X, y)
        best_model.save("cnn_augmentation_step0.8.h5")
        print("\nmodelul optim a fost salvat in cnn_augmentation_step0.8.h5")
    else:
        print("nu s-au putut incarca datele.")
```
